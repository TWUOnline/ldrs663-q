## 

validity  
the main extent to which a concept, conclusion, or measurement is
well-founded and likely corresponds accurately to the real world.

reliability  
the overall consistency of a measure. A measure is said to have a high
reliability if it produces similar results under consistent conditions

bias  
a systematic tendency in which the methods used to gather data and
estimate a sample statistic present an inaccurate, skewed or distorted
depiction of reality.
